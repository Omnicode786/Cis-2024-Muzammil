```python
w1i = 0.1
w2i = 0.1
alpha = 0.1
bias = 1
k = []
y = [0,0,0,0]
x1 = [0,0,1,1]
x2 = [0,1,0,1]
z = [1,1,1,0]
D = [a - b for a, b in zip(z, y)]
threshold = 0.5
print(D)
```

```python
errorExists = True
iter = 0
while errorExists:
    errorExists = False
    iter += 1
    for i in range(len(x1)): 
        
        k = (w1i * x1[i]) + (w2i * x2[i]) + bias
        
        if k > threshold:
            this_y = 1
        else:
            this_y = 0
            
        y[i] = this_y 
            
        diff = z[i] - this_y
        print("Iteration: ", iter)
        print(round(w1i, 2), round(w2i, 2))
        if diff != 0:
            errorExists = True 
            
            w1i = w1i + (alpha * diff * x1[i])
            w2i = w2i + (alpha * diff * x2[i])
            

D = [a - b for a, b in zip(z, y)]
print("Final Weight w1 =", round(w1i, 2), "w2 =", round(w2i, 2))
print("Final Predictions", y)
print("Number of iterations it took place was: ", iter)
```
